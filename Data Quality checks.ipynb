```python
source_file = "/Volumes/namaste_catalog/vineetdb/testvolume/source_files/customers-100.csv"
source_df = spark.read.format('csv').option('header', True).option('inferSchema', True).load(source_file)
source_schema = source_df.schema
source_df.printSchema()

'''
As we can see the column names are improper and the data types are also not proper therefore it is good idea for the 
using these case for Data Quality Rules
1. column order
2. Null  order check
3. column data type
4. column size  expected and acutall coming 
5.  Success case if olumn and their datatype is matching properly
'''
```

```python
for filed in source_df.schema:
  print(filed.name)
source_cols = [field.name for field in source_df.schema.fields]
source_types = [filed.dataType for filed in source_df.schema.fields]
source_null = [field.nullable for field in source_df.schema.fields]
#source_null
```

```python
from pyspark.sql.types import *

target_order_schema = StructType([
    StructField("index", StringType(), True),
    StructField("customer_id", StringType(), True),
    StructField("first_name", StringType(), True),
    StructField("last_name", StringType(), True),
    StructField("company", StringType(), True),
    StructField("city", StringType(), True),
    StructField("country", StringType(), True),
    StructField("phone_1", StringType(), True),
    StructField("phone_2", StringType(), True),
    StructField("email", StringType(), True),
    StructField("subscription_date", StringType(), True),
    StructField("website", StringType(), True)
])
```

```python
target_order_schema.fields
```

```python
target_cols= [field.name for field in target_order_schema.fields]
target_types =[field.dataType for field in target_order_schema.fields]
target_null  =[field.nullable for field in target_order_schema.fields]

  
    
```

```python
def schema_check(source_schema , target_schema):
    if source_schema == target_order_schema:
        print('Schema match successfull')
    print('--------------------- ** ---------------------')        
    print('Schems Failure checks ')
    print('--------------------- ** ---------------------')
    cols = set(source_cols) - set(target_cols)
    if cols:
        print('Extra column mismatch', len(source_cols) - len(target_cols))
    cols = set(target_cols) -set(source_cols)
    if cols:
        print(f'under flow columns  mismatch', len(target_cols) - len(source_cols) ) 
    print('--------------------- ** ---------------------')        
    for source_type,target_type in zip(source_types,target_types):
        if source_type != target_type:
            print('Data type mismatch')
    for s,t in zip(source_null,target_null):
        if s != t:
            print('Null check mismatch')
    print('--------------------- ** ---------------------')        
    
```

```python
schema_check(source_schema , target_order_schema)
```
