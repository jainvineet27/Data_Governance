```python
dbutils.widgets.text("source",'')
source= dbutils.widgets.get("source")
print(f'source path is >>> {source}')
```

```python
df_bronze_orders= spark.read.option("multiLine",True).json(f"{source}/orders")
#display(df_bronze_orders)
df_bronze_orders.write.format("delta").mode("overwrite").saveAsTable("namaste_catalog.jobs.bronze_orders")
```

```python
df_customers = spark.read.option('multiLine',True).format('json').load(f"{source}/customers")
df_customers.write.format("delta").mode('overwrite').option("mergeSchema",True).saveAsTable('namaste_catalog.jobs.customers')
#display(df_customers)
```

```python
df_status = spark.read.option("multiLine",True).json(f"{source}/status")
df_status.write.format("delta").mode("overwrite").option("mergeSchema",True).saveAsTable("namaste_catalog.jobs.status")
#display(df_status)
```
