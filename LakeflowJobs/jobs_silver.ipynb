```python
df_orders=  spark.table("namaste_Catalog.jobs.bronze_orders")
display(df_orders)
```

```python
from pyspark.sql import functions as f 
df_orders= df_orders.withColumn("order_date", f.col("order_date").cast("date"))
df_orders.write.format("delta").mode("overwrite").saveAsTable("namaste_catalog.jobs.orders_silver")
```

```python
df_customers = spark.read.table("namaste_catalog.jobs.customers")
df= df_customers.withColumn("updated_at" , f.col("updated_at").cast("date"))
df.write.format("delta").mode("overwrite").saveAsTable("namaste_catalog.jobs.customer_silver")


```

```python
result = df.join(df_orders.alias("o"), df.customer_id == f.col("o.customer_id"), how="inner")
result = result.select("o.city", "o.customer_id", "customer_name", "email", "updated_at", "amount", "order_date", "order_id", "profit")
#display(result)
#result.write.mode("overwrite").format("delta").saveAsTable("namaste_catalog.jobs.order_wit_customers")

```
