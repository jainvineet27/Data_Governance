```sql
create or replace  temporary  view  orders_temp as
select *
from read_files('/Volumes/namaste_catalog/vineetdb/testvolume/customers-100.csv', 
  header =>True, 
  inferSchema => True, 
  format => 'csv' 
  )

```

```sql
%python 
new_cols = [ cols.replace(' ' ,'_').lower() for cols in spark.sql('select * from  orders_temp').columns ]
spark.sql('select * from  orders_temp').toDF(*new_cols).selectExpr("customer_id","first_name", "last_name", "city","country").createOrReplaceTempView("final")
#display(df)
#another option df.toDF(*nnewcols)
```

```sql
create or replace table namaste_catalog.lakeflowconnect.order_ex1 
as select * from final  ;

```
