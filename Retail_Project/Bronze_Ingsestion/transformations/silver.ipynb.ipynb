```python
dbutils.widgets.text("catalog","namastecatalog")
dbutils.widgets.text("schema","vineetdb")
dbutils.widgets.text("file_path","/Volumes/namastecatalog/vineetdb/")
dbutils.widgets.text("checkpoint_path","/Volumes/namastecatalog/vineetdb/checkpoint_location")
```

```python
catalog = dbutils.widgets.get("catalog")
schema  = dbutils.widgets.get("schema")
checkpoint  = dbutils.widgets.get("checkpoint_path")
file_path  = dbutils.widgets.get("file_path")
print(catalog)
print(schema)
print(checkpoint)
print(file_path)
```

```python
from pyspark.sql.functions import col , explode_outer ,current_date
orders_online_df=  spark.sql(f"select * from {catalog}.{schema}.orders_on")

orders_online_df = orders_online_df.withColumn("ignestion_timestamp", current_date())\
     .withColumn("file_name", col("_metadata.file_path")) \
     .withColumn("items_exp", explode_outer(col("items"))).selectExpr("customer.*","items_exp.*" , "order_id",  "order_timestamp", "payment_mode  ", "ignestion_timestamp", "file_name")


display(orders_online_df)

    







```
